```python
import os
import pandas as pd
import shap


file_root_shapley = os.path.join(
    "..", "..", "data", "exact", "shapley_value_of_myerson_coalitions_chain.csv"
)
file_root_grabisch = os.path.join(
    "..", "..", "data", "exact", "grabisch_of_myerson_coalitions_chain.csv"
)

index = [
    "P1",
    "P2",
    "P3",
    "P4",
    "P5",
    "P6",
    "P7",
    "P8",
    "P9",
    "P10",
]
shapley_only = pd.read_csv(
    file_root_shapley,
    sep=";",
    header=None,
    names=["value"]
)
shapley_only["player"] = index
psi_1 = shapley_only.set_index("player", verify_integrity=True)

r_1 = pd.read_csv(
    file_root_grabisch,
    sep=";",
    header=0
)
r_1.columns = index
r_1["player"] = index
r_1 = r_1.set_index("player", verify_integrity=True)
```

```python
positive_alpha = 0.0
negative_alpha = 0.0
positive_beta = 0.0
negative_beta = 0.0
```

```python
def filter_to_adjacency_matrix(matrix, positive_beta, negative_beta):
    def filter_edge(x):
        if x > 0 and abs(x) > positive_beta:
            return 1
        elif x < 0 and abs(x) > negative_beta:
            return 1
        else:
            return 0
    
    return matrix.copy().applymap(filter_edge)

def filter_node_list(lst, positive_alpha, negative_alpha):
    def filter_node(x):
        if positive_alpha == 0 and negative_alpha == 0:
            return True
        elif x > 0 and abs(x) > positive_alpha:
            return True
        elif x < 0 and abs(x) > negative_alpha:
            return True
        else:
            return False
    return filter(filter_node, lst)
```

```python
A = filter_to_adjacency_matrix(r_2, positive_beta, negative_beta).to_numpy()
M = r_2.copy().to_numpy()
```

```python
# ============================================================================
# NEW CLUSTERING METHOD: Normalization + Louvain
# ============================================================================
# This replaces the duo_louvain method with a generalized approach:
# 1. normalize_weighted_adjacency: Handles positive/negative relations by:
#    - Computing normalization coefficients based on weight proportions
#    - Creating a fully non-negative weighted adjacency matrix
# 2. detect_communities: Applies Louvain algorithm on the normalized matrix
#
# Note: For this dataset, resolution=1.0 produces the same communities as
#       the original duo_louvain method (2 communities: SM vs LG/MD airports)
# ============================================================================

from cgt_perezsechi.modeling.cluster import (
    normalize_weighted_adjacency,
    detect_communities
)

# Normalize the adjacency matrix (r_1 contains positive and negative relations)
M_normalized = normalize_weighted_adjacency(
    r_1.to_numpy(), 
    alpha1=0.5,  # Weight for positive relations
    alpha2=0.5   # Weight for negative relations
)

# Detect communities using Louvain algorithm
columns = r_1.columns.tolist()
clusters = detect_communities(
    M_normalized,
    nodes=columns,
    resolution=1,  # resolution=1.0 matches duo_louvain results (0.65 gives 1 community)
    seed=42          # For reproducibility
)
```
